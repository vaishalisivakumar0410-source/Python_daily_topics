```python
matrix = [
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
]
print(matrix[0])
print(matrix[0][0])
print(matrix[1][1])
print(len(matrix))
```

```python
for i in matrix:
    print(i)
```

```python
for i in matrix:
    for j in i:
        print(j)
```

```python
for row in matrix:
    for value in row:
        print(value)
```

```python
arr = []
for row in matrix:
    for value in row:
        arr.append(value)
print(arr)  
```

```python
for row in range(len(matrix)):
    for col in range(len(matrix[row])):
            # print(matrix[row][col])
            arr.append(matrix[row][col])
print(arr)
        
```
