```python
def factorial(n:int) -> int:
    fact_value = 1
    for i in range(1, n+1):
        fact_value *= i
    return fact_value
result= factorial(5)
print(result)
```

```python
def sum(n: int) -> int:
    sum_value = 0
    for i in range(n+1):
        sum_value += i
    return sum_value
res = sum(20)
print(res)
```
