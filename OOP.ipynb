```python
class Human:
    def __init__(self,name,age):
        print(self)
        self.name = name
        self.age = age
    def greeting(self):
        print(f"Hi {self.name}, I'm {self.age} years old")
h1 = Human("Gopal", 26)
print(f"From outside of class {h1}")
h1.greeting()
print(h1.name, h1.age)
h2 = Human("Divya", 24)
print(f"From outside if class {h2}")
h2.greeting()
print(h2.name, h2.age)
```

```python
# polymorphism
class Employee:
    def __init__(self,name,age,salary):
        self.name = name
        self.age = age
        self.salary = salary

    def __add__(self, other):
        print("Addition Operator called in employee class")
        # return f"First emp {self.name} other emp {other.name}"
        # return f"{self.age + other.age}"
        return f"{self.salary + self.salary}"

e1 = Employee("Gopal", 45, 20000)
e2 = Employee("Mani", 25, 60000)
e3 = e1 + e2
print(e3)
```

```python
# abstraction
from abc import ABC, abstractmethod

class RBI_bank(ABC):
    @abstractmethod
    def interestRate(self):
        pass

class Axis(RBI_bank):
    def interestRate(self):
        print("Axis Bank Interest Rate : 8.59%")

class Sbi(RBI_bank):
    def interestRate(self):
        print("SBI Bank Interest Rate : 6.59%")

c1 = Axis()
c1.interestRate()

c2 = Sbi()
c2.interestRate()
```
